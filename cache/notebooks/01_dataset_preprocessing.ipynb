# Notebook 1: ZINC-250K Preprocessing & Scaffold Splitting
**Project:** Reinforcement Learning for Drug-Likeness Optimization of Small Molecules
**Course:** 22AIE401 – Reinforcement Learning

### Objective
1. Inspect the ZINC-250K dataset.
2. Recompute genuine QED and SAS scores using RDKit.
3. Partition the dataset using Bemis-Murcko scaffolds into 80% Train, 10% Validation, and 10% Test splits (zero scaffold overlap).
4. Stratify test molecules into 4 QED headroom groups: Low, Medium, High, and Very High.

```
import os, sys, json
import pandas as pd
import matplotlib.pyplot as plt
sys.path.insert(0, os.path.abspath('../..'))
from cache.implementation.data_pipeline import prepare_scaffold_splits
ds_path = os.path.join('../..', 'ZINC-250K', '250k_rndm_zinc_drugs_clean_3.csv')
out_dir = os.path.join('../..', 'cache', 'results')
train_df, val_df, bench_df = prepare_scaffold_splits(ds_path, out_dir, sample_size=1000)
print('Benchmark molecules sample:')
bench_df[['smiles', 'qed', 'sas', 'qed_group']].head()
```

```
plt.figure(figsize=(8, 4))
plt.hist(train_df['qed'], bins=25, alpha=0.7, color='steelblue', label='Train QED')
plt.hist(bench_df['qed'], bins=25, alpha=0.7, color='coral', label='Test Benchmark QED')
plt.title('QED Distribution across Train & Test Benchmark')
plt.xlabel('QED')
plt.ylabel('Count')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
