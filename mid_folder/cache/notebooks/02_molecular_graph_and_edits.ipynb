# Notebook 2: Molecular Graph Representation & Valence Masked Edits

### Objective
1. Convert SMILES to PyTorch Geometric `Data` graphs with 21 node features and 4 edge features.
2. Demonstrate valid atomic additions, removals, substitutions, and bond changes.
3. Illustrate chemical valence pre-action masking across the 1901 discrete action vocabulary.

```
import os, sys, torch
from rdkit import Chem
from rdkit.Chem import Draw
sys.path.insert(0, os.path.abspath('../..'))
from cache.implementation.graph_utils import mol_to_pyg_data
from cache.implementation.action_handler import get_action_mask, action_to_details, apply_action
from cache.implementation.action_handler import TOTAL_ACTIONS

smiles = 'CC(=O)Oc1ccccc1C(=O)O' # Aspirin
mol = Chem.MolFromSmiles(smiles)
data = mol_to_pyg_data(mol, current_step=0, max_steps=10, qed=0.55, sas=1.6)
print(f'PyG Data: Nodes x={data.x.shape}, Edge Index={data.edge_index.shape}, Edge Attr={data.edge_attr.shape}')
mask = get_action_mask(mol)
print(f'Total Actions: {TOTAL_ACTIONS}, Valid Actions: {mask.sum()}')
Draw.MolToImage(mol)
```
